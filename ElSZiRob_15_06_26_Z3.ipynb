```python
from keras.datasets import mnist
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
```

```python
from keras import models
from keras import layers
```

```python
siec=models.Sequential()
siec.add(layers.Dense(512,activation='relu',input_shape=(28*28,)))
siec.add(layers.Dense(10,activation='softmax'))
```

```python
siec.compile(optimizer='rmsprop',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
train_images=train_images.reshape((60000,28*28))
train_images=train_images.astype('float32')/255

test_images=test_images.reshape((10000,28*28))
test_images=test_images.astype('float32')/255
```

```python
train_images[1]
```

```python
test_labels[3]
```

```python
from keras.utils import to_categorical

train_labels = to_categorical(train_labels)
test_labels = to_categorical(test_labels)
```

```python
siec.fit(train_images,train_labels,epochs=5,batch_size=64)
```

```python
test_loss, test_acc = siec.evaluate(test_images, test_labels)
print('test_acc', test_acc)
```
